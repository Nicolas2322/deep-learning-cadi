```python
# ============================================================
# SEMANA 15 - ACTIVIDAD 15
# DATA JOURNEY + WEIGHTS & BIASES + MODEL SERVING
# ============================================================

# ============================================================
# INSTALACIÓN DE LIBRERÍAS
# ============================================================

# Instalación de Weights & Biases
!pip install -q wandb

# ============================================================
# IMPORTACIÓN DE LIBRERÍAS
# ============================================================

# TensorFlow para Deep Learning
import tensorflow as tf

# NumPy para operaciones matemáticas
import numpy as np

# Matplotlib para visualizaciones
import matplotlib.pyplot as plt

# Weights & Biases para monitoreo
import wandb

# Callback oficial de W&B para TensorFlow
from wandb.integration.keras import WandbMetricsLogger

# Métricas y herramientas de sklearn
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report

# ============================================================
# INICIALIZACIÓN DE WEIGHTS & BIASES
# ============================================================

# Inicializa un nuevo experimento en W&B
wandb.init(
    project="Semana15-DeepLearning",
    name="FashionMNIST-CNN",
    config={
        "epochs": 10,
        "batch_size": 64,
        "learning_rate": 0.001,
        "optimizer": "adam"
    }
)

# ============================================================
# CARGA DEL DATASET
# ============================================================

# Fashion MNIST contiene imágenes de ropa en escala de grises
# Se divide automáticamente en entrenamiento y prueba

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

# ============================================================
# VISUALIZACIÓN DEL DATASET
# ============================================================

# Mostrar algunas imágenes del dataset
plt.figure(figsize=(10,10))

for i in range(9):

    # Crear subgráficas
    plt.subplot(3,3,i+1)

    # Mostrar imagen
    plt.imshow(x_train[i], cmap='gray')

    # Mostrar etiqueta
    plt.title(f"Clase: {y_train[i]}")

    # Quitar ejes
    plt.axis('off')

plt.suptitle("Ejemplos del Dataset Fashion MNIST")
plt.show()

# ============================================================
# PREPROCESAMIENTO DE LOS DATOS
# ============================================================

# Normalización:
# Convertimos los valores de píxeles de rango 0-255
# a un rango entre 0 y 1 para mejorar el entrenamiento

x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

# ============================================================
# AJUSTE DE DIMENSIONES
# ============================================================

# Las CNN requieren un canal adicional
# Dimensión final:
# (cantidad_imágenes, alto, ancho, canales)

x_train = np.expand_dims(x_train, axis=-1)
x_test = np.expand_dims(x_test, axis=-1)

# ============================================================
# INFORMACIÓN DEL DATASET
# ============================================================

print("Forma de x_train:", x_train.shape)
print("Forma de x_test:", x_test.shape)

# ============================================================
# CONSTRUCCIÓN DEL MODELO CNN
# ============================================================

# Sequential permite construir el modelo capa por capa
model = tf.keras.Sequential()

# ============================================================
# PRIMERA CAPA CONVOLUCIONAL
# ============================================================

# Conv2D:
# Extrae características espaciales de las imágenes

model.add(
    tf.keras.layers.Conv2D(
        filters=32,
        kernel_size=(3,3),
        activation='relu',
        input_shape=(28,28,1)
    )
)

# ============================================================
# MAXPOOLING
# ============================================================

# Reduce dimensionalidad espacial
model.add(tf.keras.layers.MaxPooling2D((2,2)))

# ============================================================
# SEGUNDA CAPA CONVOLUCIONAL
# ============================================================

model.add(
    tf.keras.layers.Conv2D(
        filters=64,
        kernel_size=(3,3),
        activation='relu'
    )
)

# ============================================================
# SEGUNDO MAXPOOLING
# ============================================================

model.add(tf.keras.layers.MaxPooling2D((2,2)))

# ============================================================
# FLATTEN
# ============================================================

# Convierte mapas de características en vector plano
model.add(tf.keras.layers.Flatten())

# ============================================================
# CAPA DENSA
# ============================================================

# Aprende patrones complejos
model.add(tf.keras.layers.Dense(128, activation='relu'))

# ============================================================
# CAPA DE SALIDA
# ============================================================

# 10 neuronas = 10 clases
model.add(tf.keras.layers.Dense(10, activation='softmax'))

# ============================================================
# RESUMEN DEL MODELO
# ============================================================

model.summary()

# ============================================================
# COMPILACIÓN DEL MODELO
# ============================================================

model.compile(

    # Optimizador Adam
    optimizer='adam',

    # Función de pérdida para clasificación multiclase
    loss='sparse_categorical_crossentropy',

    # Métrica principal
    metrics=['accuracy']
)

# ============================================================
# ENTRENAMIENTO DEL MODELO
# ============================================================

history = model.fit(

    # Datos de entrenamiento
    x_train,
    y_train,

    # Número de épocas
    epochs=10,

    # Tamaño de lote
    batch_size=64,

    # Datos de validación
    validation_data=(x_test, y_test),

    # Callback para monitoreo en W&B
    callbacks=[WandbMetricsLogger()]
)

# ============================================================
# EVALUACIÓN DEL MODELO
# ============================================================

loss, accuracy = model.evaluate(x_test, y_test)

print("\nResultados Finales")
print("---------------------")
print(f"Loss final: {loss:.4f}")
print(f"Accuracy final: {accuracy:.4f}")

# ============================================================
# GRÁFICA DE ACCURACY
# ============================================================

plt.figure(figsize=(10,5))

plt.plot(history.history['accuracy'], label='Entrenamiento')
plt.plot(history.history['val_accuracy'], label='Validación')

plt.title("Accuracy del Modelo")
plt.xlabel("Épocas")
plt.ylabel("Accuracy")

plt.legend()
plt.grid()

plt.show()

# ============================================================
# GRÁFICA DE LOSS
# ============================================================

plt.figure(figsize=(10,5))

plt.plot(history.history['loss'], label='Entrenamiento')
plt.plot(history.history['val_loss'], label='Validación')

plt.title("Loss del Modelo")
plt.xlabel("Épocas")
plt.ylabel("Loss")

plt.legend()
plt.grid()

plt.show()

# ============================================================
# PREDICCIONES
# ============================================================

predictions = model.predict(x_test)

# Obtener clase con mayor probabilidad
predicted_classes = np.argmax(predictions, axis=1)

# ============================================================
# MATRIZ DE CONFUSIÓN
# ============================================================

cm = confusion_matrix(y_test, predicted_classes)

print("\nMatriz de Confusión")
print(cm)

# ============================================================
# REPORTE DE CLASIFICACIÓN
# ============================================================

print("\nReporte de Clasificación")
print(classification_report(y_test, predicted_classes))

# ============================================================
# VISUALIZACIÓN DE PREDICCIONES
# ============================================================

plt.figure(figsize=(12,12))

for i in range(9):

    plt.subplot(3,3,i+1)

    plt.imshow(x_test[i].reshape(28,28), cmap='gray')

    plt.title(
        f"Real: {y_test[i]} | Pred: {predicted_classes[i]}"
    )

    plt.axis('off')

plt.suptitle("Predicciones del Modelo")
plt.show()

# ============================================================
# GUARDADO DEL MODELO
# ============================================================

# Guardar modelo entrenado
model.save("modelo_fashion_mnist.h5")

print("\nModelo guardado correctamente.")

# ============================================================
# APROXIMACIÓN A MODEL SERVING
# ============================================================

print("""
======================================================
APROXIMACIÓN A MODEL SERVING
======================================================

Este modelo podría desplegarse mediante:

1. API REST usando Flask o FastAPI
2. Aplicación Web
3. Docker Container
4. Google Cloud Run
5. AWS SageMaker
6. Azure Machine Learning
7. TensorFlow Serving

El archivo .h5 generado puede reutilizarse
posteriormente para inferencia en producción.
======================================================
""")

# ============================================================
# FINALIZAR SESIÓN W&B
# ============================================================

wandb.finish()
```

# Importancia del Monitoreo en Deep Learning

El monitoreo permite detectar problemas como:

## Overfitting

Cuando el modelo aprende demasiado bien los datos de entrenamiento pero falla al generalizar.

## Underfitting

Cuando el modelo no logra aprender correctamente los patrones del dataset.

## Problemas de convergencia

Cuando el loss no disminuye adecuadamente.

## Problemas de estabilidad

Cuando las métricas presentan comportamientos irregulares.

Gracias al monitoreo continuo es posible tomar decisiones sobre:

- Ajuste de hiperparámetros.
- Cambio de arquitectura.
- Incremento de datos.
- Uso de regularización.
- Aplicación de Data Augmentation.

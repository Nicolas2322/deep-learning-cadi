```python
"""
================================================================================
SEMANA 8 - ACTIVIDAD 8
IMPLEMENTACIÓN CNN + TRANSFER LEARNING (TENSORFLOW / KERAS)
================================================================================

OBJETIVO:
---------
Implementar una red CNN para clasificación de imágenes y compararla con
Transfer Learning para analizar mejora en desempeño.

DATASET:
--------
CIFAR-10 (10 clases de imágenes RGB 32x32)

================================================================================
"""

# =========================
# 1. IMPORTACIÓN LIBRERÍAS
# =========================
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical

from sklearn.metrics import classification_report, confusion_matrix


# =========================
# 2. CARGA DEL DATASET
# =========================
def load_data():
    """
    Carga CIFAR-10 desde Keras.
    """
    (X_train, y_train), (X_test, y_test) = cifar10.load_data()
    return X_train, y_train, X_test, y_test


# =========================
# 3. PREPROCESAMIENTO
# =========================
def preprocess_data(X_train, X_test, y_train, y_test):
    """
    Normaliza imágenes y convierte etiquetas a one-hot encoding.
    """

    # Normalización: escala píxeles entre 0 y 1
    X_train = X_train.astype("float32") / 255.0
    X_test = X_test.astype("float32") / 255.0

    # One-hot encoding de clases
    y_train = to_categorical(y_train, 10)
    y_test = to_categorical(y_test, 10)

    return X_train, X_test, y_train, y_test


# =========================
# 4. MODELO CNN BASE
# =========================
def build_cnn_model():
    """
    CNN desde cero:
    - Conv2D + ReLU
    - MaxPooling
    - Flatten
    - Dense
    """

    model = models.Sequential([
        layers.Conv2D(32, (3,3), activation='relu', input_shape=(32,32,3)),
        layers.MaxPooling2D(2,2),

        layers.Conv2D(64, (3,3), activation='relu'),
        layers.MaxPooling2D(2,2),

        layers.Conv2D(128, (3,3), activation='relu'),

        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model


# =========================
# 5. TRANSFER LEARNING
# =========================
def build_transfer_model():
    """
    Modelo con Transfer Learning usando MobileNetV2 preentrenado.
    """

    base_model = MobileNetV2(
        input_shape=(32,32,3),
        include_top=False,
        weights='imagenet'
    )

    base_model.trainable = False  # congelar capas

    model = models.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model


# =========================
# 6. ENTRENAMIENTO
# =========================
def train_model(model, X_train, y_train):
    """
    Entrena modelo CNN o Transfer Learning.
    """

    history = model.fit(
        X_train, y_train,
        epochs=10,
        batch_size=64,
        validation_split=0.2,
        verbose=1
    )

    return history


# =========================
# 7. EVALUACIÓN COMPLETA
# =========================
def evaluate_model(model, X_test, y_test, name="Modelo"):
    """
    Evalúa modelo con métricas completas y matriz de confusión.
    """

    loss, acc = model.evaluate(X_test, y_test, verbose=0)
    print(f"\n===== {name} =====")
    print(f"Accuracy: {acc:.4f}")
    print(f"Loss: {loss:.4f}")

    y_pred = np.argmax(model.predict(X_test), axis=1)
    y_true = np.argmax(y_test, axis=1)

    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))

    cm = confusion_matrix(y_true, y_pred)

    plt.figure(figsize=(6,5))
    sns.heatmap(cm, annot=False, cmap="Blues")
    plt.title(f"Matriz de Confusión - {name}")
    plt.show()


# =========================
# 8. GRÁFICAS DE ENTRENAMIENTO
# =========================
def plot_history(history, title):
    """
    Visualiza accuracy y loss.
    """

    plt.figure()

    plt.plot(history.history['accuracy'], label='train')
    plt.plot(history.history['val_accuracy'], label='val')

    plt.title(f"Accuracy - {title}")
    plt.legend()
    plt.show()

    plt.figure()

    plt.plot(history.history['loss'], label='train')
    plt.plot(history.history['val_loss'], label='val')

    plt.title(f"Loss - {title}")
    plt.legend()
    plt.show()


# =========================
# 9. PIPELINE PRINCIPAL
# =========================
def main():

    # Cargar datos
    X_train, y_train, X_test, y_test = load_data()

    # Preprocesamiento
    X_train, X_test, y_train, y_test = preprocess_data(
        X_train, X_test, y_train, y_test
    )

    # =========================
    # CNN BASE
    # =========================
    cnn_model = build_cnn_model()
    history_cnn = train_model(cnn_model, X_train, y_train)

    plot_history(history_cnn, "CNN Base")
    evaluate_model(cnn_model, X_test, y_test, "CNN Base")

    # =========================
    # TRANSFER LEARNING
    # =========================
    tl_model = build_transfer_model()
    history_tl = train_model(tl_model, X_train, y_train)

    plot_history(history_tl, "Transfer Learning")
    evaluate_model(tl_model, X_test, y_test, "Transfer Learning")


# =========================
# EJECUCIÓN
# =========================
if __name__ == "__main__":
    main()
```

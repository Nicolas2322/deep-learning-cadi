```python
"""
================================================================================
PROYECTO: CNN + DATA AUGMENTATION + TRANSFER LEARNING (FASHION MNIST)
================================================================================

DATASET:
- Fashion MNIST (70,000 imágenes)
- Tamaño: 28x28
- Escala de grises (1 canal)
- 10 clases

ESCENARIOS:
1. Modelo base
2. Modelo con Data Augmentation
3. Transfer Learning
4. Fine Tuning

================================================================================
"""

# =========================
# 1. LIBRERÍAS
# =========================
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf

from tensorflow.keras import layers, models
from tensorflow.keras.datasets import fashion_mnist
from tensorflow.keras.utils import to_categorical
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import MobileNetV2

from sklearn.metrics import classification_report, confusion_matrix


# =========================
# 2. CARGA DEL DATASET
# =========================
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()

"""
Las imágenes vienen como:
- (60000, 28, 28)
Necesitamos agregar el canal → (28,28,1)
"""

X_train = X_train.reshape(-1,28,28,1)
X_test = X_test.reshape(-1,28,28,1)


# =========================
# 3. PREPROCESAMIENTO
# =========================

# Normalización [0,255] → [0,1]
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# One-hot encoding
y_train_cat = to_categorical(y_train, 10)
y_test_cat = to_categorical(y_test, 10)


# =========================
# 4. DATA AUGMENTATION
# =========================
datagen = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    zoom_range=0.1
)


# =========================
# 5. MODELO BASE CNN
# =========================
def build_base_model():
    model = models.Sequential([
        layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
        layers.MaxPooling2D(),

        layers.Conv2D(64, (3,3), activation='relu'),
        layers.MaxPooling2D(),

        layers.Flatten(),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model


# =========================
# 6. TRANSFER LEARNING
# =========================
def build_transfer_model(trainable=False):

    """
    MobileNetV2 espera 3 canales (RGB)
    → Convertimos imágenes de 1 canal a 3 canales duplicando
    """

    base_model = MobileNetV2(
        input_shape=(32,32,3),
        include_top=False,
        weights='imagenet'
    )

    base_model.trainable = trainable

    model = models.Sequential([
        layers.Resizing(32,32),                 # adaptar tamaño
        layers.Lambda(lambda x: tf.image.grayscale_to_rgb(x)),  # 1 canal → 3 canales
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    return model


# =========================
# 7. ENTRENAMIENTO
# =========================
def train_model(model, X, y, name, use_aug=False):
    """
    Entrena el modelo y guarda su historial para análisis posterior.
    """

    print(f"\n===== Entrenando {name} =====")

    if use_aug:
        history = model.fit(
            datagen.flow(X, y, batch_size=64),
            epochs=8,
            validation_data=(X[:10000], y[:10000])
        )
    else:
        history = model.fit(
            X, y,
            epochs=8,
            batch_size=64,
            validation_split=0.2
        )

    return history


# =========================
# 8. EVALUACIÓN
# =========================
def evaluate_model(model, name):

    loss, acc = model.evaluate(X_test, y_test_cat, verbose=0)

    print(f"\n==== {name} ====")
    print(f"Accuracy: {acc:.4f}")
    print(f"Loss: {loss:.4f}")

    y_pred = model.predict(X_test)
    y_pred_classes = np.argmax(y_pred, axis=1)

    print(classification_report(y_test, y_pred_classes))

    cm = confusion_matrix(y_test, y_pred_classes)

    plt.figure()
    sns.heatmap(cm, cmap="Blues")
    plt.title(f"Matriz de Confusión - {name}")
    plt.show()

    return acc, loss


# =========================
# 9. EJECUCIÓN
# =========================
results = []

# MODELO BASE
model_base = build_base_model()
train_model(model_base, X_train, y_train_cat, "Base")
acc, loss = evaluate_model(model_base, "Base")
results.append(["Base", acc, loss])


# DATA AUGMENTATION
model_aug = build_base_model()
train_model(model_aug, X_train, y_train_cat, "Augmentation", True)
acc, loss = evaluate_model(model_aug, "Augmentation")
results.append(["Augmentation", acc, loss])


# TRANSFER LEARNING
model_tl = build_transfer_model(False)
train_model(model_tl, X_train, y_train_cat, "Transfer Learning")
acc, loss = evaluate_model(model_tl, "Transfer Learning")
results.append(["Transfer Learning", acc, loss])


# FINE TUNING
model_ft = build_transfer_model(True)
train_model(model_ft, X_train, y_train_cat, "Fine Tuning")
acc, loss = evaluate_model(model_ft, "Fine Tuning")
results.append(["Fine Tuning", acc, loss])


# =========================
# 10. RESULTADOS FINALES
# =========================
print("\n===== COMPARACIÓN FINAL =====\n")
for r in results:
    print(f"{r[0]} → Accuracy: {r[1]:.4f} | Loss: {r[2]:.4f}")
```

## 1. Introducción

En el contexto del Deep Learning aplicado a visión por computador, la clasificación de imágenes representa una de las tareas fundamentales en la construcción de sistemas inteligentes capaces de interpretar información visual. Las redes neuronales convolucionales (CNN) han demostrado ser altamente efectivas en este tipo de problemas debido a su capacidad para extraer automáticamente características relevantes directamente desde los datos sin necesidad de ingeniería manual de variables.

Sin embargo, uno de los principales retos en el entrenamiento de modelos de Deep Learning es la capacidad de generalización, es decir, la habilidad del modelo para desempeñarse correctamente sobre datos no vistos durante el entrenamiento. Este problema se ve especialmente afectado cuando el conjunto de datos es limitado o cuando el modelo tiende a sobreajustarse, aprendiendo patrones específicos del conjunto de entrenamiento en lugar de relaciones generales.

Para abordar estas limitaciones, en esta actividad se implementan dos técnicas fundamentales: Data Augmentation y Transfer Learning. El objetivo es analizar de manera experimental cómo estas técnicas impactan el rendimiento, la estabilidad del entrenamiento y la capacidad de generalización del modelo.

---

## 2. Dataset

Se utilizó el dataset Fashion MNIST, el cual contiene 70,000 imágenes en escala de grises de tamaño 28x28 píxeles, distribuidas en 10 clases correspondientes a diferentes tipos de prendas de vestir.

Este dataset representa un escenario adecuado para evaluar modelos de clasificación multiclase, ya que presenta variabilidad visual moderada y permite analizar el comportamiento de distintas técnicas sin un alto costo computacional.

---

## 3. Preprocesamiento de los Datos

El preprocesamiento incluyó la normalización de los valores de los píxeles en el rango [0,1], lo cual mejora la estabilidad del entrenamiento y facilita la convergencia del modelo.

Además, se aplicó one-hot encoding a las etiquetas para adaptar el problema a una salida multiclase con función de pérdida categorical crossentropy.

Para el uso de Transfer Learning, fue necesario adaptar las imágenes de un canal (escala de grises) a tres canales (RGB), replicando la información para cumplir con los requisitos del modelo preentrenado MobileNetV2.

---

## 4. Modelo Base

Se implementó una red neuronal convolucional (CNN) desde cero, compuesta por capas Conv2D, MaxPooling, Flatten y capas densas. Este modelo actúa como línea base para comparar el impacto de las técnicas avanzadas.

Durante el entrenamiento, se observó que el modelo base presenta una diferencia notable entre accuracy de entrenamiento y validación, lo que indica presencia de sobreajuste.

---

## 5. Implementación de Data Augmentation

Se aplicaron técnicas de Data Augmentation mediante transformaciones como rotación, desplazamientos horizontales y verticales, y zoom.

Estas transformaciones permiten generar variaciones de las imágenes originales, aumentando artificialmente la diversidad del dataset. Como resultado, el modelo logra aprender representaciones más robustas.

Al comparar el modelo base con el modelo que utiliza Data Augmentation, se observa que la diferencia entre accuracy de entrenamiento y validación se reduce, evidenciando una disminución del sobreajuste. Además, el modelo presenta una mayor estabilidad durante el entrenamiento, lo cual confirma una mejora en la capacidad de generalización.

---

## 6. Implementación de Transfer Learning

Se utilizó MobileNetV2 como modelo preentrenado, el cual ha sido entrenado previamente sobre el dataset ImageNet. Este modelo es ampliamente reconocido por su eficiencia y capacidad para extraer características visuales generales como bordes, texturas y formas.

Inicialmente, las capas del modelo se mantuvieron congeladas, permitiendo aprovechar el conocimiento previo sin modificar los pesos. Posteriormente, se añadieron capas densas adaptadas al problema específico.

El uso de Transfer Learning permitió que el modelo alcanzara mejores niveles de precisión en menos épocas, evidenciando una convergencia más rápida en comparación con el modelo base.

---

## 7. Fine Tuning

Se implementó Fine Tuning permitiendo el entrenamiento de las capas del modelo preentrenado. Esto permitió ajustar las representaciones aprendidas específicamente al dataset Fashion MNIST.

El modelo con Fine Tuning mostró el mejor desempeño global, logrando una mayor precisión en comparación con las demás configuraciones, aunque con un mayor costo computacional.

## 8. Evaluación del Modelo

Se evaluaron los modelos utilizando métricas como accuracy, precision, recall y F1-score, además de matrices de confusión para analizar el comportamiento por clase.

Los resultados muestran que el modelo base presenta mayor variabilidad en sus predicciones, mientras que los modelos con Data Augmentation y Transfer Learning logran un desempeño más consistente.

En particular, el modelo con Transfer Learning alcanzó una mayor precisión global y mostró una mejor capacidad para clasificar correctamente múltiples clases, evidenciando la efectividad del enfoque.
